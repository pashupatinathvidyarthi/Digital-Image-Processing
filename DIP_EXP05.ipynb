#### Aim:- To study and perform non-linear neighborhood operations on images:   (A.)Min 
          (B.)Max 
          (C.)Median 
          (D.)With Various padding operations(zero, 1's, line,mirror)

```python
## Min Filtering
### Import Important libraries
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

##Read image using Pillow Library
img1=Image.open("baboon.png")

## Resize image to 224x224
img1=img1.resize((224,224))

##Apply min filtering
new_image1=img1.filter(ImageFilter.MinFilter(size=3))

## Plot original image and min filtering image
fig=plt.figure(figsize=(6,4))
fig.add_subplot(1,2,1)
plt.imshow(img1)
plt.axis("off")
plt.title("Original Image")

fig.add_subplot(1,2,2)
plt.imshow(new_image1)
plt.axis("off")
plt.title("Min Filtered Image")
plt.show()
```

```python
## Minimum Filtering Operation with Different Padding

### Import Important libraries
import cv2
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

##Read image using Pillow Library
img1=cv2.imread("baboon.png")
img1=cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

## Resize image to 224x224
img1=cv2.resize(img1,(224,224))

##Apply Padding (Zero padding, Mirror Padding and Replicate Padding)
img1_padd1=cv2.copyMakeBorder(img1,20,20,20,20,cv2.BORDER_CONSTANT,value=0)
img1_padd2=cv2.copyMakeBorder(img1,20,20,20,20,cv2.BORDER_REFLECT)
img1_padd3=cv2.copyMakeBorder(img1,20,20,20,20,cv2.BORDER_REPLICATE)


##Convert array as image for the respective padded image
img1_new1=Image.fromarray(img1_padd1)
img1_new2=Image.fromarray(img1_padd2)
img1_new3=Image.fromarray(img1_padd3)

##Apply Min Filtering
result1=img1_new1.filter(ImageFilter.MinFilter(size=3))
result2=img1_new2.filter(ImageFilter.MinFilter(size=3))
result3=img1_new3.filter(ImageFilter.MinFilter(size=3))

##Plot all the results
fig=plt.figure(figsize=(15,10))
fig.add_subplot(3,3,1)
plt.imshow(img1)
plt.axis("off")
plt.title("Original Image")

fig.add_subplot(3,3,2)
plt.imshow(img1_padd1)
plt.axis("off")
plt.title("Zero Padding")

fig.add_subplot(3,3,3)
plt.imshow(img1_padd2)
plt.axis("off")
plt.title("Mirror Padding")

fig.add_subplot(3,3,4)
plt.imshow(img1_padd3)
plt.axis("off")
plt.title("Replicate Padding")

fig.add_subplot(3,3,5)
plt.imshow(result1)
plt.axis("off")
plt.title("Zero Padding with Min Filtering Image")

fig.add_subplot(3,3,6)
plt.imshow(result2)
plt.axis("off")
plt.title("Mirror Padding with Min Filtering Image")

fig.add_subplot(3,3,7)
plt.imshow(result3)
plt.axis("off")
plt.title("Replicate Padding with Min Filtering Image")

```

```python
## Max Filtering
### Import Important libraries
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

##Read image using Pillow Library
img1=Image.open("baboon.png")

## Resize image to 224x224
img1=img1.resize((224,224))

##Apply min filtering
new_image1=img1.filter(ImageFilter.MaxFilter(size=3))

## Plot original image and min filtering image
fig=plt.figure(figsize=(6,4))
fig.add_subplot(1,2,1)
plt.imshow(img1)
plt.axis("off")
plt.title("Original Image")

fig.add_subplot(1,2,2)
plt.imshow(new_image1)
plt.axis("off")
plt.title("Max Filtered Image")
plt.show()
```
