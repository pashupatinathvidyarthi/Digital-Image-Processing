#### Aim:- To study and perform non-linear neighborhood operations on images:   (A.)Min 
          (B.)Max 
          (C.)Median 
          (D.)With Various padding operations(zero, 1's, line,mirror)

```python
## Min Filtering
### Import Important libraries
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

##Read image using Pillow Library
img1=Image.open("baboon.png")

## Resize image to 224x224
img1=img1.resize((224,224))

##Apply min filtering
new_image1=img1.filter(ImageFilter.MinFilter(size=3))

## Plot original image and min filtering image
fig=plt.figure(figsize=(6,4))
fig.add_subplot(1,2,1)
plt.imshow(img1)
plt.axis("off")
plt.title("Original Image")

fig.add_subplot(1,2,2)
plt.imshow(new_image1)
plt.axis("off")
plt.title("Min Filtered Image")
plt.show()
```
